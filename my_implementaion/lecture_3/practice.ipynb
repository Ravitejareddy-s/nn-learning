```python
  import matplotlib.pyplot as plt
  
  def bar(t, labels=None):
      t = t.detach()                      # drop grad tracking if present
      plt.figure(figsize=(10, 3))
      plt.bar(range(len(t)), t.tolist())  # matplotlib wants plain Python numbers
      if labels is not None:
          plt.xticks(range(len(t)), labels)
      plt.show()
  
```

```python
x=torch.tensor([2.0,4.0])
```

```python
torch.softmax(x*2,0)*100
```

```python
bar(torch.softmax(x,0))
```

```python
bar(torch.softmax(x+100,0))
```
