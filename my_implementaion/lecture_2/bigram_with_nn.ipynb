```python
import torch
import numpy as np
import matplotlib.pyplot as plt
```

```python
names = open('names.txt','r').read().splitlines()
names = ['.'+i+'.' for i in names]
biagrams = []
for i in names:
    for j in range(len(i)-1):
        biagrams.append(i[j]+i[j+1])

stoi = []
for i in names:
    for j in i:
        if j not in stoi:
            stoi.append(j)
rstoi = {}
for num,i in enumerate(sorted(set(stoi))):
    rstoi[i] = num

inp,out = [],[]
for i in biagrams:
    inp.append(rstoi[i[0]])
    out.append(rstoi[i[1]])
```

```python
import torch.nn.functional as F 
onehot = []
for i in inp:
    t = F.one_hot(torch.tensor(i),num_classes=27).to(torch.float32)
    onehot.append(t)

weights = torch.randn(27,27,requires_grad=True)

loss = torch.tensor(0.0)
for index,i_onhot in enumerate(onehot):
    mlp_out = i_onhot.reshape(1,-1)@weights
    mlp_out = mlp_out.squeeze(0)
    exp = torch.exp(mlp_out)
    sum = torch.sum(exp)
    average = exp/sum
    pick = torch.log(average[out[index]]) * -1
    loss += pick
loss = loss/len(onehot)
```
