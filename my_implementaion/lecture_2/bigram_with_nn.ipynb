```python
import torch
import numpy as np
import matplotlib.pyplot as plt
import torch.nn.functional as F 
```

```python
torch.manual_seed(42)           # set seed for reproducibility
g = torch.Generator().manual_seed(2147483647)
```

```python
names = open('names.txt','r').read().splitlines()
names = ['.'+i+'.' for i in names]
biagrams = []
for i in names:
    for j in range(len(i)-1):
        biagrams.append(i[j]+i[j+1])

stoi = []
for i in names:
    for j in i:
        if j not in stoi:
            stoi.append(j)
rstoi = {}
for num,i in enumerate(sorted(set(stoi))):
    rstoi[i] = num

inp,out = [],[]
for i in biagrams:
    inp.append(rstoi[i[0]])
    out.append(rstoi[i[1]])
```

```python
inp
```

```python


onehot = F.one_hot(torch.tensor(inp),num_classes=27).to(torch.float32)
torch.manual_seed(42)           
g = torch.Generator().manual_seed(2147483647)
weights = torch.randn(27,27,generator=g,requires_grad=True)
```

```python
for i in range(10000):
    logits = onehot@weights
    exp = torch.exp(logits)
    denom = exp.sum(1,keepdim=True)
    probs = exp/denom
    target_probs = probs[torch.arange(len(probs)),out]
    nulls = torch.log(target_probs)*-1
    loss = nulls.mean()
    loss.backward()
    with torch.no_grad():
        weights -= (weights.grad * 0.1)
    weights.grad.zero_()
```

```python

master_numnames=[]
for i in range(50):
    num_name = []
    intial = 0
    while True:
        logits = F.one_hot(torch.tensor(intial),num_classes=27).reshape(1,-1).to(torch.float32)@weights
        exp = torch.exp(logits)
        denom = exp.sum(1,keepdim=True)
        probs = exp/denom
        # index = probs.argmax().item()
        index = torch.multinomial(probs, num_samples=1).item()
        # print(index)
        if index == 0 or len(num_name) > 50:
            break 
        else:
            intial = index 
            num_name.append(index)
    master_numnames.append(num_name)

itos = {}
for i in rstoi:
    num = rstoi[i]
    itos[num] = i 
for i in master_numnames:
    s = ''
    for j in i:
        s += itos[j]
    print(s)

```

```python
names = open('names.txt','r').read().splitlines()
names = ['..'+i+'..' for i in names]
biagrams = []
for i in names:
    for j in range(len(i)-2):
        biagrams.append(i[j]+i[j+1]+i[j+2])

inp = [i[:-1] for i in biagrams]
inp = list(set(inp))

rstoi = {}
for num,i in enumerate(sorted(set(inp))):
    rstoi[i] = num


single_str_to_int = {}
for num,i in enumerate(sorted(set([i[-1] for i in biagrams]))):
    single_str_to_int[i] = num 


inp,out = [],[]
for i in biagrams:
    inp.append(rstoi[i[:-1]])
    out.append(single_str_to_int[i[-1]])


```

```python


onehot = F.one_hot(torch.tensor(inp),num_classes=628).to(torch.float32)
torch.manual_seed(42)           
g = torch.Generator().manual_seed(2147483647)
weights = torch.randn(628,27,generator=g,requires_grad=True)
```

```python
for i in range(600):
    logits = onehot@weights
    exp = torch.exp(logits)
    denom = exp.sum(1,keepdim=True)
    probs = exp/denom
    target_probs = probs[torch.arange(len(probs)),out]
    nulls = torch.log(target_probs)*-1
    loss = nulls.mean()
    loss.backward()
    with torch.no_grad():
        weights -= (weights.grad * 50)
    weights.grad.zero_()
    print(loss)
```

```python
itos = {}
temp = 0
for i in rstoi:
    itos[temp] = i 
    temp += 1



sitos = {}
for i in single_str_to_int:
    num = single_str_to_int[i]
    sitos[num] = i 
```

```python

master_numnames=[]
for i in range(50):
    num_name = []
    intial = 0
    while True:
        logits = F.one_hot(torch.tensor(intial),num_classes=628).reshape(1,-1).to(torch.float32)@weights
        exp = torch.exp(logits)
        denom = exp.sum(1,keepdim=True)
        probs = exp/denom
        # index = probs.argmax().item()
        index = torch.multinomial(probs, num_samples=1).item()
        # print(index)
        temp = itos[intial][-1]+sitos[index]
        if index == 0 or len(num_name) > 50 or temp not in rstoi:
            break 
        else:
            intial = rstoi[temp] 
            num_name.append(index)
    master_numnames.append(num_name)


for i in master_numnames:
    s = ''
    for j in i:
        s += sitos[j]
    print(s)

```
