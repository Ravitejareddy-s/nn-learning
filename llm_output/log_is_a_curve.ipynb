# Is `x` vs `log(x)` a straight line?

**No.** It's a curve. Below: the log curve on its own, then the 3 cases so you can see *when* logs actually straighten something.

```python
import numpy as np
import matplotlib.pyplot as plt

# --- Plot 1: y = log(x) vs x  (the thing you asked about) ---
x = np.linspace(0.05, 20, 400)
y = np.log(x)

# a real straight line for comparison, matched to pass through same 2 endpoints
line = np.interp(x, [x[0], x[-1]], [y[0], y[-1]])

plt.figure(figsize=(7,4.5))
plt.plot(x, y, lw=2.5, label='y = log(x)  (what you plotted)')
plt.plot(x, line, '--', color='gray', label='an actual straight line')
plt.axhline(0, color='k', lw=0.6)
plt.title('x  vs  log(x)  ->  this is a CURVE, not a line')
plt.xlabel('x'); plt.ylabel('log(x)')
plt.legend(); plt.grid(alpha=0.3)
plt.tight_layout(); plt.show()
print('Notice: log(x) shoots up fast near 0, then bends over and flattens.')
print('The dashed straight line only touches it at the two ends.')
```

## When logs DO give a straight line

Logs straighten a curve only when there's an **exponent to undo**.
- Left: an exponential `y = 2^x`. Log the **y-axis** -> straight.
- Right: a power law `y = x^3`. Log **both** axes -> straight.

```python
fig, ax = plt.subplots(1, 2, figsize=(11,4.2))

# Exponential: y = 2^x  ->  log(y) vs x is straight
x = np.linspace(0, 8, 200)
y = 2.0**x
ax[0].plot(x, np.log(y), lw=2.5, color='tab:orange')
ax[0].set_title('Exponential y = 2^x\nplot log(y) vs x  ->  STRAIGHT')
ax[0].set_xlabel('x'); ax[0].set_ylabel('log(y)')
ax[0].grid(alpha=0.3)

# Power law: y = x^3  ->  log(y) vs log(x) is straight
x = np.linspace(1, 50, 200)
y = x**3
ax[1].plot(np.log(x), np.log(y), lw=2.5, color='tab:green')
ax[1].set_title('Power law y = x^3\nplot log(y) vs log(x)  ->  STRAIGHT')
ax[1].set_xlabel('log(x)'); ax[1].set_ylabel('log(y)')
ax[1].grid(alpha=0.3)

plt.tight_layout(); plt.show()
print('Both straight -- because log peeled off the exponent (x in 2^x, and the 3 in x^3).')
print('log(x) by itself has no exponent to peel, so it stays curved.')
```
